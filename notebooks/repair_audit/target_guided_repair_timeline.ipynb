# 원하는 글자 선택 → 2×2 생성 → 획별 목표 가이드 보정

**이전 실험 결과 ZIP은 필요 없습니다.** 기존 Drive의 획 모델 best.pt와 기준 글자·기준 획 데이터를 사용합니다.
GPU(T4 가능)에서 코드 셀 **1→8** 실행하세요.
전처리와 선택 확인만 할 때는 CPU 런타임에서 1→4까지만 실행해도 됩니다.

1. 설치 → 2. 데이터 목록 → 3. 선택용 전처리 → 4. 글자 선택 → 5. 모델 로딩 → 6. 생성·보정 → 7. 단계별 보기 → 8. ZIP 다운로드.
3번에서 로컬 자료·썸네일을 한 번 준비합니다. 4번의 검색/미리보기에서는 원본 파일을 읽지 않습니다.
다른 글자를 시험할 때는 **4번의 선택을 바꾸고 6번을 다시 실행**하면 됩니다. 모델은 다시 로딩하지 않습니다.
매번 별도 실행 폴더를 만들어 기존 결과를 보존합니다.

기본: 선택한 2×2 시트 × 시드 100/101 = 생성 2장 + 정상 시트 1장, 총 3장 보정.
생성할 입력 폰트와 보정 기준 획은 따로 선택합니다. 한 실험의 같은 글자는 기준 한 종류를 공유합니다.
기준은 실행 전에 변경할 수 있으며 실행 때 복사한 조건은 이전 결과와 함께 보존됩니다.
기준을 입력 폰트에 맞춰 자동 선택하지 않습니다. 원하는 글자가 여러 칸에 반복돼도 됩니다.
글자 이름이 없으면 썸네일과 row 번호로 고릅니다. 숫자 ID를 한글 문자로 추측하지 않습니다.

```
# 4. 전처리된 그림으로 선택. 원본 파일 읽기와 썸네일 생성은 3번에서 완료합니다.
import html
from collections import Counter

def font_label(entry):
    sample = entry.get('source_entry',entry)
    path = sample.get('reference') or sample.get('source_image') or ''
    stem = Path(path.replace('\\','/')).stem
    character = entry.get('resolved_character') or reference_identity(entry)['resolved_character']
    parts = stem.split('_')
    if character in parts:
        prefix = parts[:parts.index(character)]
        if prefix[:1]==['I']:prefix=prefix[1:]
        if prefix:return '_'.join(prefix)
    return '폰트명 없음'


LABEL_TO_ENTRY = {f"row {e['manifest_row']} | {e.get('char_id') or '(이름 없음)'} | {font_label(e)}": e for e in CATALOG}
LABELS = list(LABEL_TO_ENTRY)
ROW_TO_LABEL = {e['manifest_row']: label for label, e in LABEL_TO_ENTRY.items()}
CHARACTER_COUNTS = Counter(e['resolved_character'] for e in CATALOG if e.get('resolved_character'))
_prototype_records = dict(PROTOTYPE_RECORDS)
_prototype_images = dict(PROTOTYPE_PREVIEWS)
# Preserve the existing selection contract consumed by cell 5.
reference_selectors = [widgets.Combobox(options=LABELS, value=LABELS[min(i,len(LABELS)-1)],
    ensure_option=True, description='입력 row', layout=widgets.Layout(width='285px')) for i in range(4)]
character_keys = [widgets.Combobox(options=list(_prototype_records),
    value=CATALOG[min(i,len(CATALOG)-1)].get('resolved_character') or '',description='글자',continuous_update=False,
    layout=widgets.Layout(width='285px')) for i in range(4)]
prototype_selectors = [widgets.Combobox(options=LABELS, value='', ensure_option=True, description='기준 row',
    layout=widgets.Layout(width='285px')) for _ in range(4)]
identity_checks = [widgets.Checkbox(value=False, description='입력과 기준이 같은 글자임을 확인',
    indent=False, layout=widgets.Layout(width='295px')) for _ in range(4)]
pin_buttons = [widgets.Button(description='기준 적용 / 변경', layout=widgets.Layout(width='180px')) for _ in range(4)]
slot_picker = widgets.ToggleButtons(options=[('왼쪽 위',0),('오른쪽 위',1),('왼쪽 아래',2),('오른쪽 아래',3)],
    value=0, style={'button_width':'71px'}, layout=widgets.Layout(width='305px'))
pick_mode = widgets.ToggleButtons(options=[('입력 폰트 선택','input'),('공통 기준 선택','prototype')],
    value='input', style={'button_width':'145px'}, layout=widgets.Layout(width='305px'))
search = widgets.Text(placeholder='이름 또는 row 번호',continuous_update=False,layout=widgets.Layout(width='180px'))
search_button = widgets.Button(description='검색',layout=widgets.Layout(width='60px'))
previous_page = widgets.Button(description='이전', layout=widgets.Layout(width='65px'))
next_page = widgets.Button(description='다음', layout=widgets.Layout(width='65px'))
page_label = widgets.HTML()
gallery = widgets.GridBox(layout=widgets.Layout(grid_template_columns='repeat(4, 106px)', grid_gap='6px'))
active_form = widgets.VBox(layout=widgets.Layout(width='305px'))
status = widgets.HTML()
summary = widgets.HBox(layout=widgets.Layout(flex_flow='row wrap'))
_page = 0
_thumbnail_cache = SELECTION_THUMBNAILS


def selected_rows():
    values = [w.value for w in reference_selectors]
    if any(v not in LABEL_TO_ENTRY for v in values): raise ValueError('각 칸의 입력 이미지를 선택하세요.')
    return [LABEL_TO_ENTRY[v]['manifest_row'] for v in values]


def selected_keys():
    keys = [w.value.strip() for w in character_keys]
    for i,(key, check) in enumerate(zip(keys, identity_checks)):
        if not key:raise ValueError(f'칸 {i+1}의 글자 이름을 확인하세요.')
        _,record=load_prototype(PROTOTYPE_BANK,key)
        candidate=LABEL_TO_ENTRY.get(prototype_selectors[i].value)
        if candidate and candidate['manifest_row']!=record['manifest_row']:
            raise ValueError('새로 고른 기준의 적용 버튼을 먼저 누르세요.')
        input_character=LABEL_TO_ENTRY[reference_selectors[i].value].get('resolved_character')
        base_character=reference_identity(record.get('reference',{}))['resolved_character']
        if input_character and base_character:
            if input_character!=base_character or input_character!=key:
                raise ValueError(f'칸 {i+1}의 입력과 기준 글자가 다릅니다.')
        elif not check.value:
            raise ValueError(f'칸 {i+1}의 입력과 기준이 같은 글자인지 확인하세요.')
    return keys


def _thumb(entry):
    return _thumbnail_cache[entry['manifest_row']]


def _message(text): status.value = '<span>'+html.escape(text)+'</span>'


def refresh_selection(change=None):
    index = slot_picker.value
    entry = LABEL_TO_ENTRY.get(reference_selectors[index].value)
    pictures = []
    if entry:
        pictures.append(widgets.VBox([widgets.HTML(html.escape('입력 · '+font_label(entry)+' · row '+str(entry['manifest_row']))),
            widgets.Image(value=_thumb(entry),format='png',width=92,height=92)]))
    key = character_keys[index].value.strip()
    known_character = entry.get('resolved_character') if entry else None
    character_keys[index].disabled = bool(known_character)
    character_keys[index].description = '글자(자동)' if known_character else '글자 이름'
    preview = None; caption = '기준 미선택'
    candidate = LABEL_TO_ENTRY.get(prototype_selectors[index].value)
    if candidate:
        preview = _thumb(candidate)
        caption = '선택한 기준 · '+font_label(candidate)+' · row '+str(candidate['manifest_row'])
    elif key in _prototype_images:
        preview = _prototype_images[key]
        caption = '사용할 기준 · '+font_label(_prototype_records[key].get('reference',{}))+' · row '+str(_prototype_records[key]['manifest_row'])
    if preview is not None:
        pictures.append(widgets.VBox([widgets.HTML(html.escape(caption)),
            widgets.Image(value=preview,format='png',width=92,height=92)]))
    else: pictures.append(widgets.HTML('오른쪽 그림에서<br>공통 기준을 선택하세요.'))
    prototype_entry = candidate or _prototype_records.get(key,{}).get('reference',{})
    prototype_character = reference_identity(prototype_entry)['resolved_character']
    if known_character and prototype_character:
        identity_checks[index].value = known_character==prototype_character==key
        identity_checks[index].disabled = True
        identity_checks[index].description = '동일 글자 자동 확인' if identity_checks[index].value else '입력과 기준의 글자가 다릅니다'
    else:
        identity_checks[index].disabled = False
        identity_checks[index].description = '입력과 기준이 같은 글자임을 확인'
    advanced = widgets.Accordion(children=[widgets.VBox([reference_selectors[index],prototype_selectors[index]])],
        selected_index=None,layout=widgets.Layout(width='300px'))
    advanced.set_title(0,'row 직접 입력 (선택 사항)')
    active_form.children = (widgets.HBox(pictures),character_keys[index],pin_buttons[index],
                            identity_checks[index],advanced)
    cards = []
    for i,w in enumerate(reference_selectors):
        e = LABEL_TO_ENTRY.get(w.value)
        if not e: continue
        label = ['왼쪽 위','오른쪽 위','왼쪽 아래','오른쪽 아래'][i]
        button = widgets.Button(description=label,layout=widgets.Layout(width='94px'))
        button.on_click(lambda b,i=i:setattr(slot_picker,'value',i))
        cards.append(widgets.VBox([button,widgets.Image(value=_thumb(e),format='png',width=64,height=64)],
            layout=widgets.Layout(border='2px solid #4387ee' if i==index else '1px solid #ccc',padding='3px',margin='0 8px 0 0')))
    summary.children = tuple(cards)


def choose_thumbnail(row):
    index = slot_picker.value
    label = ROW_TO_LABEL[row]
    if pick_mode.value=='input':
        reference_selectors[index].value = label
        _message(f'현재 칸의 입력을 row {row}로 선택했습니다.')
    else:
        key = character_keys[index].value.strip()
        prototype_selectors[index].value = label
        _message(f'기준 row {row} 선택. 이름을 확인하고 기준 적용을 누르세요. 실행 전에는 다시 바꿀 수 있습니다.')
    refresh_selection()


def pin(index):
    try:
        label = prototype_selectors[index].value
        if label not in LABEL_TO_ENTRY: raise ValueError('오른쪽 그림에서 공통 기준을 먼저 선택하세요.')
        key = character_keys[index].value.strip()
        record = save_prototype(CATALOG,LABEL_TO_ENTRY[label]['manifest_row'],key,PROTOTYPE_BANK)
        _prototype_records[key] = record
        _prototype_images[key] = _thumb(LABEL_TO_ENTRY[label])
        # Same character uses ONE shared choice in all slots in this experiment.
        for i,key_widget in enumerate(character_keys):
            if key_widget.value.strip()==key:
                prototype_selectors[i].value=label
                identity_checks[i].value=False
        keys = list(_prototype_records)
        for widget in character_keys: widget.options = keys
        _message(f'{record["character"]}: 기준 row {record["manifest_row"]} 적용. 필요하면 다시 선택해 변경하세요.')
    except (ValueError,FileNotFoundError) as exc: _message(str(exc))
    refresh_selection()
    render_gallery()


def render_gallery(change=None):
    global _page
    query = _applied_query
    entries = [e for label,e in LABEL_TO_ENTRY.items() if query in label.casefold()]
    source_entry = LABEL_TO_ENTRY.get(reference_selectors[slot_picker.value].value,{})
    character = source_entry.get('resolved_character')
    if pick_mode.value=='prototype' and character:
        entries = [e for e in entries if e.get('resolved_character')==character]
    if CHARACTER_COUNTS:
        entries.sort(key=lambda e:(e.get('resolved_character') or '\uffff',font_label(e),e['manifest_row']))
    pages = max(1,(len(entries)+7)//8); _page = min(_page,pages-1)
    children = []
    current_row=source_entry.get('manifest_row')
    saved_row=_prototype_records.get(character_keys[slot_picker.value].value.strip(),{}).get('manifest_row')
    for entry in entries[_page*8:(_page+1)*8]:
        row = entry['manifest_row']
        button = widgets.Button(description=f'선택 {row}',layout=widgets.Layout(width='100px',height='28px'))
        button.on_click(lambda b,row=row:choose_thumbnail(row))
        key=entry.get('resolved_character')
        count=CHARACTER_COUNTS.get(key,0)
        badges=' · '.join(name for active,name in [(row==current_row,'현재 입력'),(row==saved_row,'적용 기준')] if active)
        caption=widgets.HTML('<b>'+html.escape(key or '?')+'</b>'+ (f' · 후보 {count}개' if count else '')+
            '<br>'+html.escape(font_label(entry))+'<br>'+html.escape(badges),layout=widgets.Layout(width='98px'))
        children.append(widgets.VBox([widgets.Image(value=_thumb(entry),format='png',width=92,height=92),button,caption],
            layout=widgets.Layout(border='1px solid #ccc',padding='3px',align_items='center')))
    gallery.children = tuple(children)
    suffix = f' · {character}의 기준만 표시' if pick_mode.value=='prototype' and character else ''
    page_label.value = f'{_page+1} / {pages} · {len(entries)}개'+suffix
    previous_page.disabled = _page==0; next_page.disabled = _page==pages-1


def move_page(delta):
    global _page
    _page = max(0,_page+delta); render_gallery()


def changed(index, key_changed=False):
    identity_checks[index].value = False
    if key_changed: prototype_selectors[index].value=''
    else:
        entry=LABEL_TO_ENTRY.get(reference_selectors[index].value,{})
        character=entry.get('resolved_character')
        if character and character_keys[index].value!=character:character_keys[index].value=character
    refresh_selection()
    render_gallery()


def filter_changed(change):
    global _page, _applied_query
    _page = 0; _applied_query = search.value.strip().casefold(); render_gallery()


def mode_changed(change):
    global _page, _applied_query
    _page=0;_applied_query='';search.value='';render_gallery()


for i,widget in enumerate(reference_selectors):widget.observe(lambda change,i=i:changed(i),names='value')
for i,widget in enumerate(character_keys):widget.observe(lambda change,i=i:changed(i,True),names='value')
for widget in prototype_selectors:widget.observe(refresh_selection,names='value')
for i,button in enumerate(pin_buttons):button.on_click(lambda b,i=i:pin(i))
slot_picker.observe(refresh_selection,names='value')
slot_picker.observe(render_gallery,names='value')
pick_mode.observe(mode_changed,names='value')
_applied_query = ''
search_button.on_click(filter_changed)
previous_page.on_click(lambda b:move_page(-1)); next_page.on_click(lambda b:move_page(1))
display(widgets.VBox([
    widgets.HTML('① 칸 선택 → ② 입력/공통 기준 선택 → ③ 그림 아래 버튼 클릭. 글자 이름은 파일 정보에서 자동 연결합니다. 기준은 실행 전에 변경 가능합니다.'),
    summary,
    widgets.HBox([
        widgets.VBox([slot_picker,pick_mode,active_form],layout=widgets.Layout(width='310px',margin='0 12px 0 0')),
        widgets.VBox([widgets.HBox([search,search_button,previous_page,next_page]),page_label,gallery])
    ],layout=widgets.Layout(flex_flow='row wrap',align_items='flex-start')),
    status
]))
refresh_selection(); render_gallery()
print('선택 후 5번 모델 로딩 → 6번 실행. 다음 실험은 선택을 바꾸고 6번만 실행.')
```

```
# @title 6. 생성·보정
REPAIR_STEPS = 60 # @param {type:"integer"}
REPAIR_STRENGTH = .25
REPAIR_SEED = 2026
if not 4 <= REPAIR_STEPS <= 1000:
    raise ValueError('REPAIR_STEPS는 4~1000 사이로 설정하세요.')
print(f'DDIM 설정 {REPAIR_STEPS}스텝 × 강도 {REPAIR_STRENGTH} → 경로당 약 {int(REPAIR_STEPS*REPAIR_STRENGTH)}스텝')
# 6. 현재 선택으로 새 실행. 다른 글자를 선택한 뒤 이 셀만 다시 실행 가능
SELECTED_ROWS = selected_rows()
PROTOTYPE_KEYS = selected_keys()
OUTPUT = Path('/content')/('selected_target_repair_'+datetime.now().strftime('%Y%m%d_%H%M%S_%f'))
OUTPUT.mkdir()
SEED_CASES = prepare_selected_sheet(CATALOG,SELECTED_ROWS,OUTPUT/'selected_reference',size=CANVAS_SIZE,
    prototype_keys=PROTOTYPE_KEYS,prototype_bank=PROTOTYPE_BANK)
for condition in load_conditions(SEED_CASES[0]['directory']):
    if condition['reference'].shape[-2:] != (EXTRACTOR_SIZE,EXTRACTOR_SIZE):
        raise ValueError('선택한 기준과 모델 해상도가 다릅니다.')
SCORES = json.loads(Path(SCORES_JSON).read_text(encoding='utf-8')) if SCORES_JSON else {}
INPUT_MODE = 'fresh'
json_write(OUTPUT/'runtime_manifest.json',dict(source_hashes=SOURCE_HASHES,
    selected_rows=SELECTED_ROWS,prototype_keys=PROTOTYPE_KEYS,prototype_bank=str(PROTOTYPE_BANK),
    prototype_records=SEED_CASES[0]['prototype_records'],
    data_preparation=PREPARATION,checkpoint_sha256=file_hash(CHECKPOINT),
    manifest_sha256=file_hash(TRAIN_MANIFEST),model_id=MODEL_ID,revision=REVISION,
    scores=SCORES,gpu=torch.cuda.get_device_name(0),
    settings=dict(generation_strengths=GENERATION_STRENGTHS,generation_seeds=GENERATION_SEEDS,
        generation_steps=GENERATION_STEPS,steps=REPAIR_STEPS,strength=REPAIR_STRENGTH,
        seed=REPAIR_SEED,plan=PLAN_CONFIG)))
json_write(OUTPUT/'scheduler_config.json',SCHEDULER_CONFIG)
json_write(OUTPUT/'versions.json',VERSIONS)
print('생성용 입력 폰트（고정 기준과 별도）')
display(Image.open(Path(SEED_CASES[0]['directory'])/'input.png').resize((384,384)))
def save_snapshot():
    archive = shutil.make_archive(str(OUTPUT)+'_checkpoint', 'zip', OUTPUT)
    if DRIVE_RESULT_DIR:
        destination = Path(DRIVE_RESULT_DIR); destination.mkdir(parents=True, exist_ok=True)
        shutil.copy2(archive, destination/Path(archive).name)
    return archive

# 기존 오류 생성물이 아닌 정답 기준 시트에서 새 입력을 생성합니다.
if INPUT_MODE == 'fresh':
    generation_root = OUTPUT/'fresh_inputs'
    generation_report = generation_root/'generation_report.json'
    expected_count = sum(c['kind']=='normal_control' for c in SEED_CASES)*(1+len(GENERATION_STRENGTHS)*len(GENERATION_SEEDS))
    cached_generation = json.loads(generation_report.read_text(encoding='utf-8')) if generation_report.is_file() else None
    if cached_generation and len(cached_generation['cases']) == expected_count:
        if (cached_generation['strengths'] != GENERATION_STRENGTHS or cached_generation['seeds'] != GENERATION_SEEDS
                or cached_generation['steps'] != GENERATION_STEPS):
            raise ValueError('생성 설정이 바뀌었습니다. 2번부터 새로 실행하세요.')
        CASES = cached_generation['cases']
        print('완료된 새 생성 결과 재사용:',len(CASES))
    else:
        if generation_root.exists():
            generation_root.rename(OUTPUT/('interrupted_generation_'+datetime.now().strftime('%Y%m%d_%H%M%S_%f')))
        try:
            # 표준 img2img 입력은 fp16 UNet/text embedding dtype을 따릅니다.
            # 기존 커스텀 보정 샘플러는 이후 fp32 VAE로 돌립니다.
            pipe.vae.to(dtype=torch.float16)
            CASES = generate_fresh_cases(pipe,SEED_CASES,generation_root,
                strengths=GENERATION_STRENGTHS,seeds=GENERATION_SEEDS,steps=GENERATION_STEPS,
                on_saved=lambda case,cases: save_snapshot())
        finally:
            pipe.vae.to(dtype=torch.float32)
            save_snapshot()
    print('새 생성 + 정상 대조군:',len(CASES),'장. 이제 각 시트를 보정합니다.')

def show_comparison(case):
    root = Path(case['directory']); repair = root/'target_repair'
    panels = [('input', root/'input.png'), ('target: NOT diffusion', repair/'target_plan/aligned_target.png'),
              ('redenoise only', repair/'reverse/redenoise_only.png'),
              ('target guided', repair/'reverse/stroke_projected.png')]
    side = 288; image = Image.new('RGB',(side*4,side+30),'white'); draw = ImageDraw.Draw(image)
    for i,(label,path) in enumerate(panels):
        draw.text((i*side+5,8),label,fill='black')
        with Image.open(path) as panel: image.paste(panel.convert('RGB').resize((side,side)),(i*side,30))
    image.save(root/'diffusion_comparison.png'); display(image)

for case in CASES:
    root = Path(case['directory']); result = root/'target_repair'
    try:
        if (result/'pipeline.json').is_file():
            print('완료 사례 건너뜀:',case['case_id']); show_comparison(case); continue
        if result.exists():
            result.rename(root/('interrupted_attempt_'+datetime.now().strftime('%Y%m%d_%H%M%S_%f')))
        source = Image.open(root/'input.png').convert('RGB')
        conditions = load_conditions(root)
        masks = None  # 현재 생성물에서 새로 추출
        if masks is None:
            print(case['case_id'],'저장 추출 없음: 기존 모델로 추출',flush=True)
            try:
                extractor.to('cuda')
                masks = extract_sheet_masks(source,conditions,extractor)
            finally:
                extractor.to('cpu'); gc.collect(); torch.cuda.empty_cache()
        else: print(case['case_id'],'동일 입력의 저장 추출 재사용',flush=True)
        cache = root/'cached_extraction'; cache.mkdir(exist_ok=True)
        import numpy as np
        for i,mask in enumerate(masks,1): np.savez_compressed(cache/f'slot_{i}.npz',extracted_masks=mask)
        arm_clock = {}
        def progress(arm,step,total,row):
            now = time.monotonic()
            arm_clock.setdefault(arm,now)
            if step == 1 or step%3 == 0 or step == total:
                elapsed = now-arm_clock[arm]
                eta = elapsed/(step-1)*(total-step) if step>1 else None
                print(case['case_id'],arm,f'{step}/{total}',
                      '남은 초≈'+str(round(eta)) if eta is not None else '시간 측정 중',flush=True)
        guide, report = repair_generated_sheet(pipe,extractor,source,conditions,result,
            cached_masks=masks,scores=SCORES.get(case['case_id']),config=TargetPlanConfig(**PLAN_CONFIG),
            steps=REPAIR_STEPS,strength=REPAIR_STRENGTH,seed=REPAIR_SEED,
            prompt=case.get('prompt',''),negative_prompt=case.get('negative_prompt',''),progress=progress,save_step_images=True)
        selected = sum(s['edit_strength']>0 for s in guide.report['strokes'])
        print('수정 대상:',selected,'/',len(guide.report['strokes']), '획; 실제 스텝:',report['actual_steps'])
        show_comparison(case)
        (root/'failure.json').unlink(missing_ok=True)
    except KeyboardInterrupt:
        print('수동 중단: 부분 결과 백업. 현재 결과는 8번으로 다운로드하세요. 6번 재실행은 새로운 실험입니다.')
        break
    except Exception as exc:
        json_write(root/'failure.json',dict(error=str(exc),traceback=traceback.format_exc()))
        print('실패:',case['case_id'],str(exc),flush=True)
    finally:
        snapshot = save_snapshot()
        print('백업:',snapshot,flush=True)
        gc.collect(); torch.cuda.empty_cache()
print('현재 완료',sum((Path(c['directory'])/'target_repair/pipeline.json').is_file() for c in CASES),'/',len(CASES))
```

```
# 7. 재역확산 단계별 보기
from io import BytesIO

if 'OUTPUT' not in globals():
    raise ValueError('먼저 6번 셀을 실행하세요.')

available_runs = {}
for report_path in sorted(OUTPUT.glob('fresh_inputs/*/target_repair/reverse/report.json')):
    report = json.loads(report_path.read_text(encoding='utf-8'))
    if report.get('step_images'):
        available_runs[report_path.parents[2].name] = (report_path.parent, report)
if not available_runs:
    raise ValueError('6번 셀에서 단계 이미지가 저장된 사례가 없습니다.')

case_picker = widgets.Dropdown(options=list(available_runs), description='사례')
step_slider = widgets.IntSlider(description='단계', min=0, continuous_update=False,
                                layout=widgets.Layout(width='95%'))
frame_preview = widgets.Image(format='png')
frame_label = widgets.HTML()


def display_frame(change=None):
    reverse, report = available_runs[case_picker.value]
    frames = report['step_images']
    index = step_slider.value
    if index == 0:
        name, caption = 'input.png', '입력'
    elif index > len(frames):
        name, caption = 'stroke_projected.png', '최종 가이드 결과'
    else:
        frame = frames[index-1]
        name = frame['path']
        caption = f"스텝 {frame['step']}/{report['actual_steps']} · timestep {frame['timestep']} · 가이드 적용 후 깨끗한 이미지 예측"
    frame_label.value = caption
    preview = Image.new('RGB',(768,384),'white')
    for column,path in enumerate([reverse/name, reverse.parent/'target_plan/aligned_target.png']):
        with Image.open(path) as panel:
            preview.paste(panel.convert('RGB').resize((384,384)),(column*384,0))
    buffer = BytesIO()
    preview.save(buffer,format='PNG')
    frame_preview.value = buffer.getvalue()


def change_case(change=None):
    step_slider.value = 0
    step_slider.max = len(available_runs[case_picker.value][1]['step_images'])+1
    display_frame()


case_picker.observe(change_case,names='value')
step_slider.observe(display_frame,names='value')
change_case()
display(widgets.VBox([case_picker,step_slider,frame_label,
                      widgets.HTML('왼쪽 · 선택 단계 / 오른쪽 · 정합 목표'),frame_preview]))
```
