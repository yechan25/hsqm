```python
#@title 1. 준비하기 — ▶ 누르고 완료될 때까지 기다리세요
import sys, subprocess, tempfile
from pathlib import Path

작업폴더 = Path(tempfile.mkdtemp(prefix="hsqm_easy_", dir="/content"))
subprocess.run(["git", "clone", "--depth", "1", "https://github.com/yechan25/hsqm.git", str(작업폴더)], check=True)
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", str(작업폴더 / "requirements.txt")], check=True)
sys.path.insert(0, str(작업폴더))
# 셀을 다시 실행해도 이번에 받은 최신 코드 사용
for 이름 in list(sys.modules):
    if 이름 == "stroke_model" or 이름.startswith("stroke_model."):
        del sys.modules[이름]
print("1번 완료! 이제 2번의 ▶ 버튼을 누르세요.")
```
